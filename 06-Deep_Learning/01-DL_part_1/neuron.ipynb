```python
import torch
import torch.nn as nn
```

```python
# input
torch.manual_seed(42)
inputs=torch.tensor([1.0,2.0,3.0]) #tensor creates multidimentional array(here we created 1D)
```

```python
 neuron=nn.Linear(in_features=3,out_features=1) #activation function
```

```python
output=neuron(inputs)
```

```python
neuron.bias
```

```python
# verify ...formula
neuron.weight @ inputs + neuron.bias
```
