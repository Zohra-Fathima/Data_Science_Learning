# Anomaly detection using DBSCAN

```python
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python
X,y=make_moons(
    n_samples=500,
    noise=0.1,
    random_state=42
)
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
X_scaled.shape
```

```python
#visualize

sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1])
```

```python
#DBSCAN

dbscan=DBSCAN(
    eps=0.18,
    min_samples=5
)

labels=dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:,0],y=X_scaled[:,1],c=labels)
```

# Anomaly detection using Isolation forest

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("thyroid_dataset.csv")
```

```python
df.head()
```

```python
X=df.drop("Outlier_label",axis=1)
y=df["Outlier_label"]
```

```python
X.head()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
from sklearn.ensemble import IsolationForest

clf=IsolationForest(n_estimators=200,contamination='auto',random_state=42)
```

```python
labels=clf.fit_predict(X_scaled)
```

```python
labels
```

```python
#visualize
from sklearn.decomposition import PCA

pca=PCA(n_components=2)
X_pca=pca.fit_transform(X_scaled)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(X_pca[:,0],X_pca[:,1],c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
import numpy as np

n_outliers=np.sum(labels==-1) #checking the total number of outliers
n_normal=np.sum(labels==1)
print("outlier=",n_outliers)
print("normal=",n_normal)
```
