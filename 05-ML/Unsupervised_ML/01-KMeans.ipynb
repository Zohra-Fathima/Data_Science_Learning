```python
import os
os.environ["OMP_NUM_THREADS"] = "4" #to remove warnigs(by gpt)
import seaborn as sns
from sklearn.datasets import make_blobs
```

```python
X,y=make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,
    random_state=42
)
```

```python
X
```

```python
#visualize
sns.scatterplot(x=X[:,0],y=X[:,1])
```

# K-means clustering

```python
from sklearn.cluster import KMeans
```

```python
K=4

kMeans=KMeans(
    n_clusters=K,
    random_state=42
)
```

```python
labels=kMeans.fit_predict(X) #label = cluster number
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],c=labels)
```

## choose our K value -elbow method, silhouette score

#### elbow method

```python
wcss=[]
for k in range(1,21):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
    
```

```python
sns.lineplot(x=range(1,21),y=wcss,marker='o')
```

#### silhouette score

```python
from sklearn.metrics import silhouette_score
```

```python
ss=[]

for k in range(2,21):
    kmeans=KMeans(n_clusters=k)
    labels=kmeans.fit_predict(X)
    score=silhouette_score(X,labels)

    ss.append(score)
```

```python
#plot -k and ss

sns.lineplot(x=range(2,21),y=ss,marker='o')
```

# KMeans for iris dataset

```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris=load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
#visualize:
sns.scatterplot(x=X[:,0],y=X[:,2])
```

```python
#scaling
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
#optional - dimensionality reduction using PCA
from sklearn.decomposition import PCA

pca=PCA(n_components=2)

pca_data=pca.fit_transform(X_scaled)
```

```python
#elbow method
wcss=[]

for k in range(1,11):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit_predict(pca_data)
    wcss.append(kmeans.inertia_)

sns.lineplot(x=range(1,11),y=wcss,marker='o')
```

```python
#KMeans:

kmeans=KMeans(n_clusters=3,random_state=10)
labels=kmeans.fit_predict(pca_data)

sns.scatterplot(x=pca_data[:,0],y=pca_data[:,1],c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],marker='X',c="red",s=220)
```
