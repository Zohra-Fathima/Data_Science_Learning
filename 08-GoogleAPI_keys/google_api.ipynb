## text input => text output

```python
from dotenv import load_dotenv
import os
from google import genai

load_dotenv()

google_api_key=os.getenv("GEMINI_API_KEY")
```

```python
from google import genai

client = genai.Client(api_key=google_api_key)

interaction = client.interactions.create(
    model="gemini-3.6-flash",
    input="what is the lifespan of hen"
)

print(interaction.output_text)
```

```python
interaction = client.interactions.create(
    model="gemini-3.6-flash",
    input="a+2=7, find a",
    system_instruction="act like a math teacher"
)

print(interaction.output_text)
```

## AI personal assistant

```python
#anser queries
#summarize emails
class PersonalAssistant:
    def __init__(self):
        print("HI, I am your AI assistant, How can I help you?")

    def ans_query(self):
        question=input("ask me anything: ")
        
        interaction = client.interactions.create(
            input=question,
            model="gemini-3.6-flash",
            system_instruction="act like a helpgul personal assistant",
             generation_config={
                 "temperature":0.7,
                 "max_output_tokens":512
             }   
        )
        
        print(interaction.output_text.strip())


    def summarize_email(self):
        email_text=input("paste your email here: ")
        prompt=f"summarize the following email in 2-3 sentences:{email_text}"
        
        interaction = client.interactions.create(
            input=prompt,
            model="gemini-3.6-flash",
            system_instruction="act like an expert email assistant",
             generation_config={
                 "temperature":0.7,
                 "max_output_tokens":512
             }   
        )
        
        print("summary",interaction.output_text.strip())
```

```python
assistant=PersonalAssistant()
```

```python
assistant.ans_query()
```

```python
assistant.summarize_email()
```

```python
import os
from dotenv import load_dotenv
load_dotenv(override=True)   # re-reads the .env file

from huggingface_hub import InferenceClient

print(os.getenv("HUGGING_FACE_IMAGE_API") is not None)  # should say True

client = InferenceClient(
    api_key=os.getenv("HUGGING_FACE_IMAGE_API"),
    provider="auto"
)

image = client.text_to_image(
    "A highly detailed futuristic spaceship, flying near a nebula",
    model="black-forest-labs/FLUX.1-schnell"
)

os.makedirs("images", exist_ok=True)
image.save("images/generated_image.png")
image
```
