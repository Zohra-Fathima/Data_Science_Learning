```python
#cleaning data- handle missing values

df=pd.read_csv("raw_data.csv")

df.isnull()
df.isna()

df.isnull().sum()#counts the number of missing values per column
df.dropna() #drops the rows which have missing values
df.dropna(axis=1) #drops the columns which have missing values
df.fillna(0) #replace missing values by any value we pass

age_mean=df["age"].mean()
df["age"].fillna(age_mean)

cleaned_data=df.copy()
age_mean=df["age"].mean()
cleaned_data["age"]=cleaned_data["age"].fillna(age_mean)

cleaned_data

df.ffill() #fills forward
df.bfill() #fills backward
```

```python
#feature engineering
#we transform the data like adding new columns or adding new features like we add tax and new_income column which is new
#(refer notes by maam)

df2=df.copy()

df2["tax"]=df2["income"].apply(lambda x: "20%" if x >= 50000 else "10%")

gender_map={"Male":"M","Female":"F","Unknown":"U"}
df2["gender"]=df2["gender"].map(gender_map)

df2=df2.assign(new_income=df2["income"] * 1.1)

df2["country"]=df2["country"].replace("USA","United States")

df2.columns=["Id","Name","Age","Country","Gender","Income","Tax","New_Income"]
#rename rows or columns
df2.rename(columns={"Income":"Salary"}) #to change column's label we use columns
df2.rename(index={1:"First"}) #to change row's label we use index

#sort
df2.sort_values("Income") #sort on the basis of income
df2=df2.fillna(50)
df2.sort_values("Income",ascending=False) #sort in descending order

sorted_df=df2.sort_values(["Income","Age"]) #sorting on the basis of two columns 

sorted_df.sort_index() 


sorted_df
#reset
#after doing the above code our id name were jumbled so using sort_index() makes the index values sorted
#doing so pandas adds a new column with the index values and returns the label which is sorted

sorted_df.reset_index()
sorted_df.reset_index(drop=True)#if we dont want original index to be preserved


#Ranking
sorted_df["Ranking"]=sorted_df["Income"].rank(ascending=False,method="min") # to get whole number method="dense"
sorted_df

#reorder
sorted_df[["Id","Name","Age","Country","Gender","Income","New_Income","Tax"]] #simply pass the column values in the order you want them


```

```python
#grouping and aggregation of data;

df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].min()

df.groupby("country")["income"].agg(["mean","min","max"]) #to calculate all the three values use agg()

df.groupby("country")["income"].agg(avg_salary="mean",min_salary="min",max_salary="max")

df.groupby("country").agg({
    "income":"mean",
    "age":"mean"
    
})

df.groupby("country").agg({
    "income":"min",
    "age":"max"
    
})

df.groupby("country").agg(
    max_salary=("income","max"),
    min_age=("age","min")
    
)
```

```python
#melt and pivot

df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22],
})

melted_data=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metrics",
    value_name="value"
)

melted_data.pivot(
    index=["country","year"],
    columns="metrics",
    values="value"
)
```

```python
# Merging & Joining Data

df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})

pd.merge(df_customers, df_orders, on="customer_id") #inner join
pd.merge(df_customers, df_orders, on="customer_id",how="left") #left join
pd.merge(df_customers, df_orders, on="customer_id",how="right") #right join
pd.merge(df_customers, df_orders, on="customer_id",how="outer") #outer join: keeps values of both the tables
```
