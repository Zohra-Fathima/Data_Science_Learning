```python
#beautiful scoup

from bs4 import BeautifulSoup

with open("scraped_data/data1.html","r")as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"lxml") #beautiful soup is a class and we are creating an object 
```

```python
#finding the country and population info from countries.csv file:
soup.find("h1") #returns first occurence of element
all_h3=soup.find_all("h3")

all_countries=[]
for h3 in all_h3:
    name=(h3.get_text(strip=True)) #get_text is used to access tag's content
    # print(h3.find_parent())

    population=(h3.find_next("div").select("span.country-population")[0].get_text())

    all_countries.append([name,population])
```

```python
import pandas as pd

df_countries=pd.DataFrame(all_countries,columns=["name","population"])
```

```python
df_countries.to_csv("cleaned_data/countries.csv",index=False)
```

```python
#activity:
#fetch data nad store it
page_count=1

while True:
    URL=f"https://quotes.toscrape.com/page/{page_count}/"
    res=requests.get(URL)

    soup=BeautifulSoup(res.text,"lxml")
    quotes=soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore...")
        break

    with open(f"scraped_data/quotes{page_count}.html","w",encoding="utf-8")as f:
        f.write(res.text)
        print(f"downloaded data from page{page_count}")

    page_count=page_count+1
```

```python
#extract useful info -page1

file_count=1
life_quotes=[]

while True:

    try:
        with open(f"scraped_data/quotes{file_count}.html","r",encoding="utf-8")as f:
            html_content=f.read()
        soup=BeautifulSoup(html_content,"lxml")

        all_quotes=soup.select("div.quote")
       

        for q in all_quotes:
            all_tags=[]

            for tag in q.select(".tags .tag"):
                all_tags.append(tag.get_text())
            
        
            if "life" in all_tags:
                text=q.select_one("span.text").get_text()
                author=q.select_one("small.author").get_text()
                life_quotes.append([text,author])
        
        
    except:
        print("no more pages...")
        break
        
    
    file_count=file_count+1
for quote in life_quotes:
    print(quote[0])
    print(quote[1])
    print()
```
