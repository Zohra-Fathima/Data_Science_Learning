```python
#real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

#plot
plt.plot(years,oscar_revenue)
plt.title("Oscar movies revenue in each year")
plt.xlabel("years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years,oscar_revenue,color="red",marker="<",linestyle="-.",linewidth=5,label="oscar movies") #format strings
plt.plot(years,non_oscar_revenue,"<--y",label="non-oscar movies")

plt.title("oscar movies vs non-oscar movies")
plt.xlabel("years")
plt.ylabel("Revenue (in $M)")

plt.legend()
```

```python
plt.plot(years,oscar_revenue,label="oscar movies") #format strings
plt.plot(years,non_oscar_revenue,label="non-oscar movies")

plt.title("oscar movies vs non-oscar movies")
plt.xlabel("years")
plt.ylabel("Revenue (in $M)")

plt.legend()

# plt.style.available #returns lists of all available styles
# plt.style.use("default")
plt.grid()
plt.show()
```

```python
#bar charts

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

plt.bar(years,oscar_revenue,color="black")
plt.title("revenue in each year")
plt.xlabel("year")
plt.ylabel("revenue (in $M)")

for i in range(len(years)): #printing the exact values above each bar
    plt.text(years[i],oscar_revenue[i]+10,str(oscar_revenue[i]),ha="center")

plt.ylim(0,max(oscar_revenue)+100)
```

```python
#multiple datasets on bar graph
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = [
    "Slumdog Millionaire",
    "Avatar",
    "Inception",
    "Hugo",
    "Lincoln"
]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M
years = [2008, 2009, 2010, 2011, 2012]


import numpy as np
x=np.arange(len(years))
width=0.4

plt.bar(x-width/2,oscar_revenue,width,label="oscar movies") #we are shifting to left. But why from x? because x position's each category on the x-axis
plt.bar(x+width/2,non_oscar_revenue,width,label="non-oscar movies") #we are shifting to right

plt.title("oscar vs non-oscar movies revenue")
plt.xlabel("year")
plt.ylabel("revenue($M)")
plt.legend()


plt.xticks(x,years)
```

```python
# Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]

age = [22, 25, 30, 35, 42, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 125, 130, 135, 140, 123, 145, 150]

colors=["green" if x<135 else "red" for x in bp]

plt.scatter(age,bp,s=bp,cmap="OrRd",c=bp) #here s is size. alpha shows the transperency of markers
plt.title("age vs bp")
plt.xlabel("age")
plt.ylabel("blood pressure")

plt.grid()
plt.colorbar(label="bp")#to show the scale


#annotations
for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i],bp[i]),xytext=(age[i]+1,bp[i]+1))

plt.xlim(min(age),max(age)+10)
plt.ylim(min(bp),max(bp)+5)
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]


colors=["red","black","pink","yellow","purple"]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
       wedgeprops={
           "edgecolor":"black",
           "linewidth":1,
           "linestyle":"--",
           # "fill":False
       }) #autopct-> shows perventage of each slice
plt.show()
```
