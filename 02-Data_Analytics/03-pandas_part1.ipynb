```python
s=pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#indexing
print(s[0])
print(s[3])
```

```python
#vectorize

s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])

print(s1+s2)

#values are mutable, size is immutable
s1[0]=100
print(s1)

changed_s1=s1.drop(0)
print(changed_s1)

print(s1)
```
