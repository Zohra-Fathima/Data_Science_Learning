```python
import pandas as pd
```

```python
info={
    "name":["rahul","zohra","fathima"],
    "CGPA":[9.5,8.7,6.0]
}

df=pd.DataFrame(info)

print(df)
```

```python
s=pd.Series([1,2,3,4,5])
print(s)
print(type(s))

#indexing
print(s[0])
print(s[3])
```

```python
#label
s=pd.Series([25,23,26,27],index=["adam","bob","eve","muna"])
print(s)
print(s["adam"])
print(s.index)
```

```python
#vectorize

s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])

print(s1+s2)

#values are mutable, size is immutable
s1[0]=100
print(s1)

changed_s1=s1.drop(0)
print(changed_s1)

print(s1)
```

```python
#dataframe:-
#1=>dictionery of lists:
info={
    "name":["rahul","zohra","fathima"],
    "age":[23,24,25],
    "CGPA":[9.5,8.7,6.0]
}

df=pd.DataFrame(info)
print(df,type(df))

print(df.index)#to print rows labels use .index
print(df.columns)#to print columns labels use .columns
```

```python
#2=>list of lists:
df=pd.DataFrame([["adam",40],["bob",20],["eve",30]],columns=["name","age"]) #we should pass the headings seperately using column
print(df) 
```

```python
#3=>np arrays:
import numpy as np

np_arr=np.array([[1,2,3],[4,5,6]])

df=pd.DataFrame(np_arr, columns=["a","b","c"])

print(df)
```

```python
#json data
df=pd.read_json("employee_data.json") ##reads the json and converts it into a DataFrame
print(df,type(df))
```

```python
#csv data
df=pd.read_csv("employee_data.csv") #reads the CSV and converts it into a DataFrame
print(df,type(df))
```

```python
#methods in df:
#we can also pass particular values in the following functions...
df.head() #prints starting 5 rows
df.tail() #prints last 5 values
df.sample()#prints a random value
df.info() #returns summary of data
df.shape #returns rows and columns
df.describe() #perfoms and returns some operations on numerical data

df.columns#prints all the columns
df.nunique
```
